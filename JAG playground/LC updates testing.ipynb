```python
time_range = ('1988', '2023')
query = {
    'y': lat_range,
    'x': lon_range,
    'time': time_range,
    'product': product,
    'dask_chunks':{'time': 1, 'x': 2048, 'y': 2048}
}

# Load DEA Land Cover data from the datacube
lc = dc.load(
    output_crs='EPSG:3577',
    measurements=[
        'level3',
        'level4'
    ],
    resolution=(-90, 90),
    **query
)
```

```python
"""
width_pixels seems to be analagous to out_width, so they should have the same name.
width_pixels is also misleading because it will lead to users putting in large numbers.
"""

level_3 = lc["level3"]

# Generate plot
lc_animation(level_3, 
            width_pixels=3,
            label_size=10,
            font_size=4)
```

```python
# Select Level 4
level_4 = lc["level4"]

file_name = 'level4_shallow_inlet_stacked'
# Generate plot
lc_animation(level_4, 
             file_name=file_name,
             width_pixels=3, 
             label_size=10,
             font_size=4,
             stacked_plot=True)
```

```python
lc_animation(level_4, 
             measurement='lifeform', 
             width_pixels=3, 
             label_size=10,
             font_size=5,
             colour_bar=True)
```
