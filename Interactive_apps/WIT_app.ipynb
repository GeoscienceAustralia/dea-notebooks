## Background

Wetlands provide a wide range of ecosystem services including improving water quality, carbon sequestration, as well as providing habitat for fish, amphibians, reptiles and birds. 

The DEA Wetlands Insight Tool summarises how the amount of water, green vegetation, dry vegetation and bare soil varies over time within each wetland. It provides the user with the ability to compare how the wetland is behaving now with how it has behaved in the past. This allows users to identify how changes in water availability have affected the wetland. It achieves this by presenting a combined view of Water Observations from Space, Tasseled Cap Wetness and Fractional Cover measurements from the Landsat series of satellites, summarised as a stacked line plot to show how that wetland has changed over time.

## DEA Wetlands Insight Tool stacked line plot tool

This tool allows you to interactively draw a polygon for a wetland of interest and get back a stacked line plot - WIT plot - showing how the behaviour of the wetland has changed over time. To use the tool:


1. Run the `wetlandsinsighttool.wit_app()` cell below; an interactive application will appear.
2. Zoom in on the map and use the `Draw a polygon` or `Draw a rectangle` tools on the left of the map to draw a shape around your wetland area of interest.
> To keep load times reasonable, the app is restricted to areas **smaller than 2000 square kilometres** in size.

<img width="600" height="600" src="../Supplementary_data/DEA_Wetlands_Insight_Tool/DEA_WIT_app_polygon-example.PNG">

3. On the left of the map, enter the name of your wetland area into `Wetland Name`. This will be used in the title of the WIT plot. 
4. A `.csv` file will be exported with this app. Use `Output CSV` to set the name of this file. 

> **Optional:** Tick the `Figure (.png)` box to export the WIT plot as a `.png` image. Note that the `Wetland Name` value will be used as the file output name. 

> **Optional:** Upload a shapefile or GeoJSON containing wetland polygons using the `Upload` button (uploaded files should be less than 5 megabytes in size).


```python
wetlandsinsighttool.wit_app()
```

## Advanced settings

This interactive tool has some advanced features.
To access these, click on the `Advanced` tab in the menu to the left of the map to expand it:

* `Minimum Good Data (%)`: A number between 0 and 100 (e.g. `85`) indicating the minimum percentage of good quality pixels required for a satellite observation to be loaded and therefore included in the WIT plot. 
* `Override maximum size limit`: Whether to override the app's default 2000 square km area limit. This can be used to load larger wetland areas, but should be used with caution as it can lead to memory issues or crashes.

> **If running the notebook for the first time**, keep the default settings below. This will demonstrate how the analysis works and provide meaningful results.
