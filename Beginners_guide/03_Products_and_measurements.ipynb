## Background

[Digital Earth Australia (DEA)](https://www.ga.gov.au/dea) is based on the [Open Data Cube](https://www.opendatacube.org/) infrastructure, and specialises in storing remotely sensed data, particularly from Earth Observation satellites such as [Landsat](https://landsat.gsfc.nasa.gov/) and [Sentinel-2](https://www.copernicus.eu/en/about-copernicus/infrastructure/discover-our-satellites).

Data is served to the public via DEA's [STAC API](https://knowledge.dea.ga.gov.au/guides/setup/gis/stac/) (embedded within [DEA's explorer](https://explorer.dea.ga.gov.au/products)), and contains both satellite observations and derivative data "products".

These data products are often composed of a range of assets, or bands. Before running a query to load data, it is useful to know what it contains.

This notebook demonstrates several straightforward ways to inspect the product and assest information of datasets within DEA's STAC catalogue.

## Description

This notebook demonstrates how to connect to DEA's STAC catalogue and interrogate the available products and assests stored within.

Topics covered include:

* How to connect to the STAC API
* How to list all the products
* How to list all of a product's assests

***

## Setup

### Load packages

Import Python packages that are used for the analysis.

```python
import odc.stac
import pandas as pd
import pystac_client
from dea_tools.datahandling import stac_collections, stac_assets
```

### Connect to DEA's STAC API

This allows us to search for data using DEA's STAC API (refer to the [SpatioTemporal Asset Catalogue guide](https://knowledge.dea.ga.gov.au/guides/setup/gis/stac/) or [Loading data with STAC notebook](../Beginners_guide/04a_Loading_data_STAC.ipynb) for details).

To load data, we must also configure access to DEA's Amazon S3 buckets using `odc.stac.configure_s3_access`.
If launching a Dask client (e.g. `dea_tools.dask.create_local_dask_cluster()`), this is done automatically.

```python
# Configure data access
odc.stac.configure_s3_access(cloud_defaults=True, aws_unsigned=True)

# Connect to DEA STAC API
catalog = pystac_client.Client.open("https://explorer.dea.ga.gov.au/stac")
```

## Analysis

### List products

Once we have connected to DEA's STAC catalogue, users can explore the products and assests/bands stored within.

The following cell lists all products that are currently available in the DEA's STAC API `stac_collections()` function. 

Products listed under **product** in the following table represent the product options available when querying the STAC API.

The table below provides some useful information about each product, including a brief product **description**, the data's **license**, the spatial extent of the product **bbox** and the start and end dates **start_date**, if applicable.

> For a comprehensive product description and access to complete product metadata, users are directed to DEA's [Explorer](https://explorer.dea.ga.gov.au/products) and the [Knowledge Hub](https://knowledge.dea.ga.gov.au/)



```python
stac_collections(catalog=catalog)
```

## List measurements


Most products are associated with a range of available assets, or bands.
These can be individual satellite bands (e.g. Landsat's near-infrared band) or statistical product summaries.

The `stac_assets()` function can be used to interrogate the assets associated with a given product (specified by the **products** column from the table above).

For example, `ga_ls5t_ard_3` refers to the Geoscience Australia Landsat 5 Analysis-ready data Collection 3 product.

The table below includes a range of technical information about each band in the `ga_ls5t_ard_3` dataset, including any **bands** which can be used to load the data, the data type or **dtype**, and the measurement's **nodata** value.

Change the `product` name below and re-run the following cell to explore available assests associated with other products.

```python
product = "ga_ls5t_ard_3"

stac_assets(catalog, products=product)
```
