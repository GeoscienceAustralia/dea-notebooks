## Background


Similar to Digital Earth Australia (DEA), [Google Earth Engine](https://earthengine.google.com/) contains a multi-petabyte catalog of satellite and environmental data, provided in analysis ready data formats hosted in the cloud. 



## Description
This notebook demonstrates how to load and analyse Google Earth Engine data in a Jupyter Notebook:

1. Authenticate and initialise the Earth Engine Python API using `ee.Authenticate()` and `ee.Initialize`
2. Search and load raster data (name) using `ee.ImageCollection`
