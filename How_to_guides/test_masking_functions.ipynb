## Troubleshooting: Module Import Issues

If you encounter `NameError` or other import issues when testing the `masking` module:

### Problem
Python caches imported modules in memory and bytecode (`__pycache__`). Changes to source files won't be reflected until the cache is cleared.

### Solution

**Step 1: Clear Python Cache**
```bash
# Remove all __pycache__ directories
find /workspaces/dea-notebooks/Tools -type d -name "__pycache__" -exec rm -rf {} +
```

**Step 2: Restart Jupyter Kernel**
- In Jupyter: `Kernel` → `Restart Kernel`
- In VS Code: Click the restart button in the notebook toolbar

**Step 3: Re-run Import Cells**
Run the import cell again to reload the module with your latest changes.

### Alternative: Use Auto-Reload

Add this at the very beginning of your notebook to automatically reload modules:

```python
%load_ext autoreload
%autoreload 2
```

This will automatically reload all modules before executing each cell, so you don't need to restart the kernel after making changes.


````python
print("Testing valid_mask...\n\n")

# test with float data and nan nodata
float_data = np.array([1.0, 2.0, np.nan, 4.0])
mask = masking.valid_mask(float_data, np.nan)
assert mask.tolist() == [True, True, False, True]
print("valid_mask with float data works\n\n")

# test with integer data and specific nodata value
int_data = np.array([1, 2, -9999, 4])
mask = masking.valid_mask(int_data, -9999)
assert mask.tolist() == [True, True, False, True]
print("valid_mask with integer data works\n\n")

# test with no nodata
mask = masking.valid_mask(int_data, None)
assert mask.all() == True
print("valid_mask with no nodata works\n\n")

print("valid_mask works\n\n")
tested_functions.add('valid_mask')
````
